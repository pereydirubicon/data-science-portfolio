```python
def head_and_info_display(df, n=6):
    """Отображает информацию о DataFrame: первые n строк, типы данных, пропуски, дубликаты"""
    
    
    print('~' * 23 + '\nИНФОРМАЦИЯ О ДАТАФРЕЙМЕ\n' + '~' * 23)
    
    # Размер данных
    print(f'Форма данных: {df.shape[0]} строк, {df.shape[1]} столбцов')
    
    # Информация об индексе
    print("\nИНФОРМАЦИЯ ОБ ИНДЕКСЕ:")
    print("-" * 40)
    idx = df.index
    if isinstance(idx, pd.MultiIndex):
        print(f"Тип: MultiIndex\nУровни: {list(idx.names)}\nУникальных комбинаций: {idx.nunique()}")
    else:
        print(f"Тип: {type(idx).__name__}\nИмя: {idx.name}\nУникальных: {idx.nunique()}")
    
    # Проверка монотонности индекса
    if hasattr(idx, 'is_monotonic_increasing'):
        inc = idx.is_monotonic_increasing
        print(f"Индекс монотонно возрастает: {inc}")
        if not inc and hasattr(idx, 'is_monotonic_decreasing'):
            print(f"Индекс монотонно убывает: {idx.is_monotonic_decreasing}")
    
    # Список столбцов
    print(f"\nСТОЛБЦЫ ({len(df.columns)} шт.):")
    print("-" * 30)
    for i, col in enumerate(df.columns, 1):
        print(f"{i:2}. {col} ('Тип': {df[col].dtype})")
    
    # Первые n строк
    print(f"\nПервые {n} строк:")
    print("-" * 15)
    display(df.head(n))
    
    # Пропущенные значения
    print("\nПРОПУЩЕННЫЕ ЗНАЧЕНИЯ:")
    print("-" * 30)
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    if missing.empty:
        print('Пропущенных значений нет')
    else:
        print(f'Столбцы с пропусками ({len(missing)} из {df.shape[1]}):')
        display(pd.DataFrame({
                                'Столбец': missing.index,
                                'Кол-во': missing.values,
                                '%': (missing.values / len(df) * 100).round(2)
                            }))
        total = missing.sum()
        cells = len(df) * df.shape[1]
        print(f'Всего пропусков: {total} ({total/cells*100:.2f}%)')
```

### Декомпозиция: Тренды и сезонность 
    
*— для определения пиковых часов, дней недели и общей динамики спроса разложим ряд на три составляющие:*

- `Тренд` — плавное изменение среднего значения
- `Сезонность` — циклически повторяющиеся закономерности
- `Остаток` (шум) — случайные колебания
    
`Декомпозиция позволяет отделить долгосрочную динамику от регулярных колебаний и случайных отклонений, что помогает понять структуру данных.`

```python
# Так как данные почасовые (суточный цикл) берем period=24 часа

# Декомпозиция ряда
decomposed = seasonal_decompose(data['num_orders'], model='additive', period=24)

# Визуализация компонент
plt.figure(figsize=(12, 8))

plt.subplot(411)
plt.plot(data.index, data['num_orders'], label='Исходный ряд', color='#F2F2A0')
plt.legend()

plt.subplot(412)
decomposed.trend.plot(ax=plt.gca(), color='#A0F2F2')
plt.title('Тренд')

plt.subplot(413)
decomposed.seasonal.plot(ax=plt.gca(), color='#F2A0F2')
plt.title('Сезонность')

plt.subplot(414)
decomposed.resid.plot(ax=plt.gca(),color='#A0A0F2')
plt.title('Остаток (шум)')

plt.tight_layout()
plt.show()
```

<div style="background-color: #FCFCF5; border: 0px solid #F2D580; border-radius: 10px; padding: 15px 25px; display: inline-block; color: black;">

- **Тренд** (бирюзовый) — долгосрочное направление изменения ряда. Здесь заметен плавный рост значений в течение рассматриваемого периода - количество заказов растёт с марта по сентябрь, что говорит о положительной динамике без учёта краткосрочных колебаний.

### Создание признаков для модели
    
---
    
ВРЕМЕННЫЕ ПРИЗНАКИ
    
---
    
1. **hour**

- Заказы такси имеют ярко выраженный суточный цикл. `[Ночной пик (00:00–04:00), Вечерний пик (16:00–23:00), Утренний спад (05:00–07:00)]`
    
2. **day_of_week** `[Уже создали]`
- Спрос зависит от дня недели:
    
    - `[Понедельник и Пятница — самые загруженные дни во все месяцы (исключение: март, где Чт и Пт равны)]`
    - `[Вторник — стабильно самый низкий спрос среди будней]`
    - `[Воскресенье — самый низкий спрос среди всех дней]` 
---    
    
ЛАГОВЫЕ ПРИЗНАКИ `[создадим в тренировочной и тестовой выборках отдельно, чтобы избежать утечки данных]` 
    
---
    
3. **lag_2**	`[2 часа назад]`
- Учёт инерции спроса (2 часа назад, возможно, тоже важно)  
4. **lag_24** `[24 часа назад (вчера в это же время)]`
- Суточный цикл — вчера в это же время спрос был похожим (утро, день, вечер, ночь)
5. **lag_168**	`[168 часов назад (ровно неделю)]`
- Недельный цикл — прошлая пятница похожа на эту пятницу, прошлый понедельник — на этот

---

```python
# Создание признаков
def create_features(df):
    # Временные признаки
    df['hour'] = df.index.hour
    # Скользящие статистики
    df['rolling_mean'] = df['num_orders'].shift().rolling(24).mean()
    
    return df
```

```python
# График автокорреляции до 170 лагов
fig, ax = plt.subplots(figsize=(24, 6))
plot_acf(data['num_orders'], lags=170, alpha=0.05, ax=ax)
ax.set_title('Автокорреляционная функция (ACF)', fontsize=16)
ax.set_xlabel('Лаг (часы)', fontsize=14)
ax.set_ylabel('Автокорреляция', fontsize=14)
ax.grid(True, alpha=0.3)
plt.show()

# Рассчитываем автокорреляцию для значимых лагов
lags = [1, 2, 3, 24, 168]
acf_values = acf(data['num_orders'], nlags=max(lags), fft=True)

# Создаём таблицу
acf_table = pd.DataFrame({
    'Лаг': lags,
    'Автокорреляция': [acf_values[lag] for lag in lags]
})

print("-" * 60)
print("АВТОКОРРЕЛЯЦИЯ ДЛЯ ЗНАЧИМЫХ ЛАГОВ")
print("-" * 60)
print(acf_table.to_string(index=False))
```

**3. Лаги 1–3 — значимы, но слабее.**

#### Количественные признаки

```python
def plot_quantitative_new(df, column, figsize=(15, 6), palette='Paired',
                          custom_labels=None, discrete_threshold=20):
    ''' 
    Ящик с усами и Гистограмма для количественных признаков
    
    Параметры:
    -----------
    df : DataFrame - Датафрейм с данными
    column : str - Название столбца для анализа
    figsize : tuple - размер области построения
    palette : str - Цветовая палитра
    custom_labels : dict - словарь с подписями
    discrete_threshold : int - порог для дискретного признака
    '''
    
    # Проверка: есть ли столбец
    if column not in df.columns:
        print(f"ОШИБКА: Столбец '{column}' не найден")
        return
    
    # Проверка: числовой ли столбец
    if not pd.api.types.is_numeric_dtype(df[column]):
        print(f"ОШИБКА: Столбец '{column}' не числовой. Тип: {df[column].dtype}")
        return
    
    # Создаём копию и приводим Int64 к int64
    data = df.copy()
    for col in data.select_dtypes(include=['Int64']).columns:
        data[col] = data[col].astype('int64')
    
    # Словарь с подписями для признаков такси
    default_labels = {
                      'num_orders':  ['Количество заказов', 'Заказы', 'Количество часов'],
                      'hour':        ['Час дня', 'Час (0-23)', 'Количество часов'],
                      'day_of_week': ['День недели', 'День (0-6)', 'Количество дней'],
                      'rolling_mean':['Скользящее среднее (24ч)', 'Среднее за 24 часа', 'Количество часов']
                      }
    
    if custom_labels:
        default_labels.update(custom_labels)
    
    labels_dict = default_labels
    
    if column in labels_dict:
        plot_title, x_label, y_label = labels_dict[column]
    else:
        plot_title = f'Распределение: {column}'
        x_label = column
        y_label = 'Количество'
    
    # Определяем дискретный признак
    unique_values = data[column].nunique()
    is_integer = data[column].dtype in ['int64', 'int32', 'int16', 'int8']
    is_discrete = (unique_values <= discrete_threshold and is_integer) or (unique_values <= 15)
    
    # Создаём два подграфика
    fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(figsize[0], figsize[1]))
    
    # Цвет для графиков
    color = sns.color_palette(palette, 1)[0]
    
    # 1. Ящик с усами
    sns.boxplot(data=data, y=column, color=color, ax=ax1)
    ax1.set_title(f'Диаграмма размаха: {plot_title}', fontsize=14)
    ax1.set_ylabel(x_label)
    
    # 2. Гистограмма
    if is_discrete:
        sns.histplot(data=data, x=column, discrete=True, color=color, ax=ax2, shrink=0.8)
        min_val = int(data[column].min())
        max_val = int(data[column].max())
        if (max_val - min_val) <= 30:
            ax2.set_xticks(range(min_val, max_val + 1))
        if unique_values <= 15:
            value_counts = data[column].value_counts().sort_index()
            for val, count in value_counts.items():
                ax2.text(val, count + max(value_counts)*0.02, f'{count}', 
                        ha='center', va='bottom', fontsize=9)
    else:
        sns.histplot(data=data, x=column, kde=True, color=color, ax=ax2)
    
    ax2.set_title(f'Гистограмма: {plot_title}', fontsize=14)
    ax2.set_xlabel(x_label)
    ax2.set_ylabel(y_label)
    
    median_val = data[column].median()
    ax2.axvline(median_val, color='red', linestyle='--', label=f'Медиана: {median_val:.1f}')
    ax2.legend()
    
    plt.tight_layout()
    plt.show()
    
    # Вывод статистики
    print(f"Статистика для '{plot_title}':")
    print('-' * 50)
    stats = data[column].describe()
    print(f"Тип данных: {'Дискретный' if is_discrete else 'Непрерывный'}")
    print(f"Уникальных значений: {unique_values}")
    print(f"Количество записей: {stats['count']:.0f}")
    print(f"Среднее: {stats['mean']:.3f}")
    print(f"Медиана: {stats['50%']:.3f}")
    print(f"Минимум: {stats['min']:.6f}")
    print(f"Максимум: {stats['max']:.3f}")
    
    # Проверка на нормальность
    if len(data[column].dropna()) >= 3:
        print(f"\nПроверка на нормальность распределения (α=0.05):")
        print('-' * 50)
        col_data = data[column].dropna()
        
        if len(col_data) <= 5000:
            from scipy.stats import shapiro
            shapiro_stat, shapiro_p = shapiro(col_data)
            print(f"Критерий Шапиро-Уилка: p-value = {shapiro_p:.4f}")
        
        try:
            from scipy.stats import normaltest
            dagostino_stat, dagostino_p = normaltest(col_data)
            if dagostino_p > 0.05:
                print('ВЫВОД: распределение соответствует нормальному')
            else:
                print('ВЫВОД: распределение НЕ соответствует нормальному')
        except Exception as e:
            print(f'Ошибка: {e}')
```

#### Корреляционный анализ

```python
def plot_phik_correlation(df, target_col, interval_cols=None, figsize=(12, 10), n_top=10):
    '''
    Строит phik матрицу корреляции с целевым признаком на первой позиции
    
    Параметры:
    -----------
    df : Исходный датафрейм
    target_col : Название целевого признака
    interval_cols : Список числовых признаков для корректной обработки
    figsize : Размер графика
    n_top : Количество топ признаков для вывода
    '''
    
    # Создаем копию данных
    df_clean = df.copy()
    
    # Исключаем ID
    if 'id' in df_clean.columns:
        df_clean = df_clean.drop('id', axis=1)
    
    # Вычисляем матрицу корреляции
    phik_matrix = df_clean.phik_matrix(interval_cols=interval_cols)
    
    # Переупорядочиваем матрицу: целевой признак на первое место
    columns_order = [target_col] + [col for col in phik_matrix.columns if col != target_col]
    phik_matrix = phik_matrix.reindex(index=columns_order, columns=columns_order)
    
    # Визуализация
    plt.figure(figsize=figsize)
    sns.heatmap(phik_matrix, 
                annot=True, 
                fmt='.2f', 
                cmap='Greens',
                center=0,
                square=True,
                cbar_kws={'shrink': 0.8})
    
    plt.title(f'Phik Correlation Matrix (Целевой признак: {target_col})', fontsize=16, pad=20)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    
    # Функция для интерпретации корреляции
    def interpret_correlation(value):
        if value < 0.3:
            return 'Слабая связь'
        elif value < 0.7:
            return 'Умеренная'
        elif value == 1:
            return 'Абсолютная'
        else:
            return 'Сильная'
    
    # топ-признаков корреляции с целевой переменной
    target_corr = phik_matrix[target_col].sort_values(ascending=False)
    
    print(f"\nТоп-{n_top} признаков по корреляции с '{target_col}':")
    print('-' * 50)
    
    # Создаем DataFrame для красивого вывода с интерпретацией
    corr_df = pd.DataFrame({
                                'Признак': target_corr.index[:n_top],
                                'Phik корреляция': target_corr.values[:n_top],
                                'Интерпретация': [interpret_correlation(x) for x in target_corr.values[:n_top]]
                            })
    
    # Форматируем вывод
    pd.set_option('display.float_format', lambda x: f'{x:.4f}')
    display(corr_df)
    
    # Выводим топ-10 коррелирующих пар признаков (без целевого)
    print(f"\nТоп-{n_top} пар признаков с наибольшей корреляцией (без '{target_col}'):")
    print("-" * 60)
    
    # Создаем список для хранения пар и их корреляций
    pairs = []
    
    # Получаем все признаки кроме целевого
    features = [col for col in phik_matrix.columns if col != target_col]
    
    # Перебираем все пары признаков
    for i in range(len(features)):
        for j in range(i + 1, len(features)):
            feat1 = features[i]
            feat2 = features[j]
            corr_value = phik_matrix.loc[feat1, feat2]
            pairs.append({
                          'Признак 1': feat1,
                          'Признак 2': feat2,
                          'Phik корреляция': corr_value,
                          'Абсолютная': abs(corr_value)
                         })
    
    # Создаем DataFrame и сортируем по абсолютной корреляции
    pairs_df = pd.DataFrame(pairs)
    pairs_df = pairs_df.sort_values('Абсолютная', ascending=False).head(n_top)
    
    # Добавляем интерпретацию
    pairs_df['Интерпретация'] = pairs_df['Phik корреляция'].apply(interpret_correlation)
    
    # Удаляем вспомогательный столбец и выводим результат
    pairs_df = pairs_df.drop('Абсолютная', axis=1)
    display(pairs_df.reset_index(drop=True))
```

#### Разделение данных на выборки

```python
# Удаляем исходные столбцы
data.drop(['day_of_week'], axis=1, inplace=True)
```

```python
# Размер тестовой выборки — 10%
test_size = int(len(data) * 0.1)

# Создаём разрыв в 168 часов (7 дней) для валидности теста с lag_168
gap = 168

# Разделяем с сохранением порядка и с разрывом
train = data.iloc[:-(test_size + gap)].copy()
test = data.iloc[-test_size:].copy()

print('-' * 50)
print('РАЗМЕР ТРЕНИРОВОЧНОЙ И ТЕСТОВОЙ ВЫБОРОК')
print('-' * 50)
print(f"Размер train: {len(train)}")
print(f"Размер test: {len(test)}")
print(f"Разрыв: {gap} часов")

# Добавляем лагированные признаки
def add_lags(df, target_col='num_orders', lags=(2, 24, 168), history=None):
    """Создаёт лаговые признаки для train или test"""
    df = df.copy()
    
    if history is not None:
        # Для test: используем историю из train
        combined = pd.concat([history[target_col], df[target_col]])
        for lag in lags:
            df[f'lag_{lag}'] = combined.shift(lag).iloc[-len(df):].values
    else:
        # Для train: просто shift
        for lag in lags:
            df[f'lag_{lag}'] = df[target_col].shift(lag)
    
    return df

# Создаём лаги для train
train = add_lags(train)
train.dropna(inplace=True)

# Создаём лаги для test с историей из train
test = add_lags(test, history=train[['num_orders']])
test.dropna(inplace=True)

print('-' * 50)
print('РАЗМЕР ВЫБОРОК ПОСЛЕ УДАЛЕНИЯ ПРОПУСКОВ')
print('-' * 50)
print(f"Размер train: {len(train)}")
print(f"Размер test: {len(test)}")

# Проверка временных границ
print('-' * 50)
print('ПРОВЕРКА ВРЕМЕННЫХ ГРАНИЦ')
print('-' * 50)
print(f"\nДата конца train: {train.index.max()}")
print(f"Дата начала test: {test.index.min()}")
print(f"Разрыв между train и test: {test.index.min() - train.index.max()}")
```

### Обучение моделей

```python
# Признаки
numeric_features = ['rolling_mean', 'hour_sin', 'hour_cos','lag_2', 'lag_24', 'lag_168']
```

```python
# Препроцессор
data_preprocessor = ColumnTransformer([
                                        ('scaler', StandardScaler(), numeric_features)
                                      ])

# Пайплайн
pipe_final = Pipeline([
                        ('preprocessor', data_preprocessor),
                        ('model', None)
                      ])

# Сетка гиперпараметров
param_grid = [
                # 1. Ridge
                {
                    'model': [Ridge(random_state=42)],
                    'model__alpha': [50.0, 80.0, 100.0, 400.0, 700.0, 1000.0],
                    'preprocessor__scaler': [StandardScaler(), MinMaxScaler(), RobustScaler(), 'passthrough']
                },

                # 2. Random Forest
                {
                    'model': [RandomForestRegressor(random_state=42)],
                    'model__n_estimators': [100,300, 500,600],
                    'model__max_depth': [2,5,7,15],
                    'preprocessor__scaler': ['passthrough']
                },
                # 3. LightGBM
                {
                    'model': [LGBMRegressor(random_state=42, n_jobs=-1, verbose=-1)],
                    'model__n_estimators': [100, 900, 1000],
                    'model__learning_rate': [0.002, 0.005, 0.01],
                    'model__max_depth': [1, 10, 15, 20, 40],
                    'preprocessor__scaler': ['passthrough']
                }
               ]

# RandomizedSearch с TimeSeriesSplit
tscv = TimeSeriesSplit(n_splits=5)

random_search = RandomizedSearchCV(
                                    pipe_final,
                                    param_grid,
                                    n_iter=30,
                                    cv=tscv, 
                                    scoring='neg_root_mean_squared_error',
                                    random_state=42,
                                    n_jobs=-1,
                                    verbose=1
                                   )

# Обучение
random_search.fit(X_train, y_train)

# Результаты
print("Лучшие параметры:", random_search.best_params_)
print("Лучший RMSE (CV):", -random_search.best_score_)
```

```python
# Границы гиперпараметров
bounds = {
                            'Ridge': {'alpha': (50.0, 1000.0)},
            
            'RandomForestRegressor': {
                                      'n_estimators': (200, 400),
                                      'max_depth': (2,15)
                                      },
                    'LGBMRegressor': {
                                      'n_estimators': (100, 1000),
                                      'learning_rate': (0.002, 0.1),
                                      'max_depth': (1, 20)
                                       }

          }
results = pd.DataFrame(random_search.cv_results_)

print("-" * 60)
print("ЛУЧШАЯ МЕТРИКА ПО КАЖДОЙ МОДЕЛИ И ПРОВЕРКА ГРАНИЦ")
print("-" * 60)

summary = []

for model_type in bounds.keys():
    # Ищем модели по имени класса
    model_results = results[results['param_model'].apply(lambda x: type(x).__name__ == model_type)]
    
    if len(model_results) == 0:
        print(f"\n[НЕТ ДАННЫХ] {model_type}")
        continue
    
    best = model_results.loc[model_results['mean_test_score'].idxmax()]
    params = best['params']
    score = -best['mean_test_score']  # так как scoring='neg_root_mean_squared_error'
    
    print(f"\n[МОДЕЛЬ] {model_type}")
    print(f"   RMSE (CV): {score:.3f}")
    
    # Проверка границ
    extreme_params = []
    for param, (min_val, max_val) in bounds[model_type].items():
        key = f'model__{param}'
        if key not in params:
            continue
        val = params[key]
        
        # Проверка на None для max_depth
        if max_val is None:
            if val == min_val:
                extreme_params.append(f"{param}={val} (мин)")
            continue
        
        if val == min_val:
            extreme_params.append(f"{param}={val} (мин)")
        elif val == max_val:
            extreme_params.append(f"{param}={val} (макс)")
    
    status = "ОПТИМУМ" if not extreme_params else "НА ГРАНИЦЕ"
    
    summary.append({
                    'Модель': model_type,
                    'RMSE (CV)': round(score, 3),
                    'Статус': status,
                    'Границы': ', '.join(extreme_params) if extreme_params else '-'
                   })
    
    # Вывод гиперпараметров лучшей модели
    print(f"   Гиперпараметры:")
    for param in bounds[model_type].keys():
        key = f'model__{param}'
        if key in params:
            print(f"     {param}: {params[key]}")

print("-" * 60)
print("ИТОГОВАЯ ТАБЛИЦА СРАВНЕНИЯ")
print("-" * 60)
print(pd.DataFrame(summary).to_string(index=False))

print("\n" + "-" * 60)
print("ВЫВОД")
print("-" * 60)

# Лучшая модель по метрике
best_row = min(summary, key=lambda x: x['RMSE (CV)'])
print(f"Лучшая модель по RMSE: {best_row['Модель']} ({best_row['RMSE (CV)']:.3f})")
```
